# DAT300 figure generation

This notebook creates the main visual outputs used in the project README.
The notebook runs the shared figure generator and saves outputs into `docs/figures/`.

```python
from pathlib import Path
import sys

workspace = Path("/workspaces")
candidates = [Path.cwd().resolve()]
if workspace.is_dir():
    candidates.extend(path for path in workspace.iterdir() if path.is_dir())

ROOT = next(
    (path for path in candidates if (path / "scripts" / "generate_figures.py").is_file()),
    None,
)
if ROOT is None:
    ROOT = next(
        (
            path.parent.parent.resolve()
            for path in workspace.rglob("generate_figures.py")
            if path.parent.name == "scripts"
        ),
        None,
    )
    if ROOT is None:
        raise FileNotFoundError(
            "Could not find scripts/generate_figures.py under /workspaces."
        )

sys.path.insert(0, str(ROOT))
from scripts.generate_figures import generate_all_figures

generate_all_figures()
```
